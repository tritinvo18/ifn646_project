```python
import os
import requests
from tqdm.notebook import tqdm

FILE_NAME = "GCF_000001405.26_GRCh38_genomic.fna.gz"
URL = "https://ftp.ncbi.nlm.nih.gov/genomes/all/GCF/000/001/405/GCF_000001405.26_GRCh38/GCF_000001405.26_GRCh38_genomic.fna.gz"
CHUNK_SIZE = 1024 * 1024  # 1 MB chunks

if os.path.exists(FILE_NAME):
    print(f"✓ {FILE_NAME} already exists. Skipping download.")
else:
    print(f"⬇ Downloading {FILE_NAME} from NCBI...")
    
    response = requests.get(URL, stream=True)
    response.raise_for_status()
    total_size = int(response.headers.get("content-length", 0))

    with open(FILE_NAME, "wb") as f, tqdm(
        desc=FILE_NAME,
        total=total_size,
        unit="iB",
        unit_scale=True,
        unit_divisor=1024,
    ) as progress_bar:
        for chunk in response.iter_content(chunk_size=CHUNK_SIZE):
            if chunk:
                size = f.write(chunk)
                progress_bar.update(size)

    print("✓ Download complete.")
```
